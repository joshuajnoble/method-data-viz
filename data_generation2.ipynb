```python
import pandas as pd

# load in the base data
base_df = pd.read_csv("superstore.csv")

# sales by year and category
segment_year_sales_df = (
    base_df
    .assign(_order_year=pd.to_datetime(base_df["Order Date"], format="%m/%d/%y").dt.year)
    .groupby(["_order_year", "Sub-Category"], as_index=False)["Sales"]
    .sum()
    .rename(columns={"_order_year": "Year","Sub-Category":"Category"})
).to_csv('apps/public/segment_year_sales_df.csv')

# sales by category/subcategory
(
    base_df.groupby(["Category", "Sub-Category"], as_index=False)["Sales"]
    .sum()
    .to_csv('apps/public/category_subcategory_sales_df.csv')
)

# sales by subcategory
(
    base_df.groupby(["Sub-Category"], as_index=False)["Sales"]
    .sum()
    .to_csv('apps/public/subcategory_sales_df.csv')
)

# aggregated metrics by category/subcategory
(
    base_df
    .groupby(["Category", "Sub-Category"], as_index=False)
    .agg(
        {
            "Sales": "sum",
            "Profit": "sum",
            "Quantity": "sum",
            "Shipping Cost": "sum",
        }
    )
    .to_csv('apps/public/category_subcategory_agg_df.csv')
)
```
